# Урок №1

```python
sells = 1000000
arenda = 200000
zarplata = 500000
komunal = 50000

rashodi = arenda + zarplata + komunal
pribil = sells - arenda - zarplata - komunal
print("Прибыль:",pribil,"\nРасходы:", rashodi)
```

# Анализ расходов


```python
import matplotlib.pyplot as plt

categories = ["Аренда", 'Зарплата', "Коммунальные услуги"]
sales = [arenda, zarplata, komunal]

plt.pie(sales, labels = categories)
plt.show()
```

#Бюджет путешествия

##1)Условия:

```python
Budget = 500000 #бюджет
Tickets = 150000 #билеты
Souvenirs = 10000 #сувениры
Excursions = 50000 #экскурсии
Sustenance = 50000 #пропитание
Housing = 100000 #жилье

The_Remaining_Budget = Budget - Tickets - Souvenirs - Excursions - Sustenance - Housing #остальной бюджет
Expenses = Tickets + Souvenirs + Excursions + Sustenance + Housing #расходы
print("Излишки:",The_Remaining_Budget,"\nРасходы:", Expenses )
```

#Анализ расходов

```python
import matplotlib.pyplot as plt

categories = ["Билеты", 'Сувениры', "Экскурсии", "Пропитание", "Жилье"]
sales = [Tickets, Souvenirs, Excursions, Sustenance, Housing ]

plt.pie(sales, labels = categories)
plt.show()
```
